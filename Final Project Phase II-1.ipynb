we need to narrow down just **foodborne records** so find out how to find that in the **primary mode** column. 

we can do this by finding 'Food' in the 'Primary Mode' column as shown above

```python
def downloaded_dataset_parser(filename):
  df = pd.read_csv(filename)

  df = df[df['Primary Mode'] == 'Food']
  df = df[['IFSAC Category', 'Food Vehicle', 'Etiology', 'Year', 'Month', 'Primary Mode', 'State', 'Illnesses']]

  # clean the unknown category labels for each food type
  df['IFSAC Category'] = df['IFSAC Category'].fillna('Unknown')

  # clean the Illness category, sees int as string becaus there is a ,
  df['Illnesses'] = df['Illnesses'].astype(str).str.replace(',', '').astype(int)

  # clean the Etiology column
  df['Etiology'] = df['Etiology'].fillna('Unknown')

  # clean the Food Vehicle na values
  df['Food Vehicle'] = df['Food Vehicle'].fillna('Unknown')

  df.to_csv('cleaned_data/downloaded/cleaned_downloaded.csv', index=False)


############ Function Call ############
downloaded_dataset_parser('data/NORS_20261101.csv')
```
