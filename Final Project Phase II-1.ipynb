```python
def downloaded_dataset_parser(filename):
  df = pd.read_csv(filename)

  df = df[df['Primary Mode'] == 'Food']
  df = df[['IFSAC Category', 'Food Vehicle', 'Etiology', 'Year', 'Month', 'Primary Mode', 'State', 'Illnesses']]

  # clean the unknown category labels for each food type
  df['IFSAC Category'] = df['IFSAC Category'].fillna('Unknown')

  # clean the Illness category, sees int as string becaus there is a ,
  df['Illnesses'] = df['Illnesses'].astype(str).str.replace(',', '').astype(int)

  # clean the Etiology column
  df['Etiology'] = df['Etiology'].fillna('Unknown')

  # clean the Food Vehicle na values
  df['Food Vehicle'] = df['Food Vehicle'].fillna('Unknown')

  df.to_csv('cleaned_data/downloaded/cleaned_downloaded.csv', index=False)


############ Function Call ############
downloaded_dataset_parser('data/NORS_20261101.csv')
```

## HTML Requirement

```python
# import statements
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin
```

```python
def html_parser():
    # STEP 1: get the webpage and find its tables
    url = 'https://www.fda.gov/food/outbreaks-foodborne-illness/investigations-foodborne-illness-outbreaks'
    response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'}, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.content, 'html.parser')
    tables = soup.find_all('table')
    records = []

    # STEP 2: get the headings and standardize their names
    for table in tables:
        rows = table.find_all('tr')

        if not rows:
            continue

        headers = []

        for cell in rows[0].find_all(['th', 'td']):
            heading = ' '.join(cell.get_text(' ', strip=True).split())
            headers.append(heading)

        # some tables use different headings for the same information
        heading_names = {'Outbreak/ Event Status': 'Outbreak/Event Status', 'Outbreak Status': 'Outbreak/Event Status', 'FDA Traceback Initiated': 'Traceback Initiated', 'FDA Inspection Initiated': 'Inspection Initiated', 'On-Site Inspection Initiated': 'Inspection Initiated', 'FDA Sampling Initiated': 'Sampling Initiated', 'Sample Collection & Analysis Initiated': 'Sampling Initiated'}
        headers = [heading_names.get(header, header) for header in headers]

        if 'Reference #' not in headers or 'Total Case Count' not in headers:
            continue # make sure that the table is an investigation table

        # STEP 3: extract each investigation and its advisory link
        for row in rows[1:]: # we already got the header so start at the 2nd row
            cells = row.find_all('td')

            if not cells:
                continue

            if len(cells) != len(headers):
                raise ValueError('The number of cells does not match the headings.')

            # remove numeric footnotes so a count like 40 does not become "40 1"
            for cell in cells:
                for footnote in cell.find_all('sup'):
                    if footnote.get_text(strip=True).isdigit():
                        footnote.decompose()

            values = []

            for cell in cells:
                text = ' '.join(cell.get_text(' ', strip=True).split())
                values.append(text)

            record = dict(zip(headers, values)) # puts the first heading with the first value, so on so forth

            count_cell = cells[headers.index('Total Case Count')]
            link = count_cell.find('a', href=True)

            if link:
                record['Case Count URL'] = urljoin(url, link['href'])
            else:
                record['Case Count URL'] = ''

            records.append(record)

    # STEP 4: create the dataframe and standardize the dates
    df = pd.DataFrame(records)
    dates = pd.to_datetime(df['Date Posted'], format='%m/%d/%Y', errors='coerce')
    missing_dates = dates.isna()

    # retry the dates with 2 digit years using their original text
    dates.loc[missing_dates] = pd.to_datetime(df.loc[missing_dates, 'Date Posted'], format='%m/%d/%y')
    df['Date Posted'] = dates

    # STEP 5: standardize the unknown descriptions
    for column in ['Pathogen or Cause of Illness', 'Product(s) Linked to Illnesses (if any)']:
        df[column] = df[column].str.strip(" '\"")
        df[column] = df[column].replace(['', 'Not Identified', 'Not Yet Identified', 'None Identified'], 'Unknown')

    # STEP 6: separate the count types and convert the counts to numbers
    df['Case Count Text'] = df['Total Case Count']
    df['Count Type'] = 'Illness cases'
    adverse_reports = df['Case Count Text'].str.contains('adverse event reports', case=False)
    df.loc[adverse_reports, 'Count Type'] = 'Adverse event reports'

    # keep the report label so we do not accidentally include these in illness totals
    count_numbers = df['Case Count Text'].str.replace(',', '', regex=False).str.replace(' adverse event reports', '', regex=False)
    df['Total Case Count'] = pd.to_numeric(count_numbers, errors='coerce').astype('Int64')
    df.loc[df['Total Case Count'].isna(), 'Count Type'] = 'See linked source'

    # an unavailable count is unknown, not zero
    df['Total Case Count'] = df['Total Case Count'].astype(object).fillna('Unknown')

    # STEP 7: clean the statuses and action indicators
    df['Investigation Status'] = df['Investigation Status'].str.split().str[0]
    df['Outbreak/Event Status'] = df['Outbreak/Event Status'].str.split().str[0]

    for column in ['Recall Initiated', 'Traceback Initiated', 'Inspection Initiated', 'Sampling Initiated']:
        df[column] = df[column].replace({'✔': 'Yes', 'See Advisory': 'Yes', '': 'Not indicated'})

    # STEP 8: export and return the cleaned dataframe
    df.to_csv('cleaned_data/html/cleaned_html.csv', index=False)
    return df


############ Function Call ############
html_parser()
```
